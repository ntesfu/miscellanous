```python
import cv2
import numpy as np
from ultralytics import YOLOE
from ultralytics.models.yolo.yoloe import YOLOEVPSegPredictor
```

```python
REF   = "/media/lm-ciss/LM_4TB/assembly_copilot/assets/test.png"
SCENE = "/media/lm-ciss/LM_4TB/assembly_copilot/aiops_parts_detection/images/PXL_20260810_061252474.jpg"

obj = cv2.imread(REF)
assert obj is not None, f"could not read {REF}"
h, w = obj.shape[:2]

x1, y1, x2, y2 = 0, 0, w, h

model = YOLOE("yoloe-11l-seg.pt")
results = model.predict(
    SCENE,
    refer_image=REF,
    visual_prompts=dict(
        bboxes=np.array([[x1, y1, x2, y2]], dtype=np.float32),
        cls=np.array([0], dtype=np.int32),
    ),
    predictor=YOLOEVPSegPredictor,
)

boxes = results[0].boxes.xyxy.cpu().numpy()
conf  = results[0].boxes.conf.cpu().numpy()
print(len(boxes), conf)
```
